## 7. Data Cleaning - Missing Values

**Output example:**
```
Original DataFrame with NaN:
   Name   Age  Salary
0  John  25.0  50000.0
1  Jane   NaN  60000.0
2   Bob  30.0     NaN

After dropna():
   Name   Age  Salary
0  John  25.0  50000.0
```

````python
# Create data with missing values
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice'],
    'Age': [25, np.nan, 30, np.nan],
    'Salary': [50000, 60000, np.nan, 55000]
})

print("DataFrame with missing values:")
print(df)

# Check for missing values
print("\nMissing values:")
print(df.isnull())

print("\nCount of missing values:")
print(df.isnull().sum())

# Drop rows with any missing value
print("\nAfter dropna():")
print(df.dropna())

# Fill missing values
print("\nAfter fillna(0):")
print(df.fillna(0))

# Fill with specific values per column
print("\nAfter fillna with mean:")
df_filled = df.copy()
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())
print(df_filled)
````

## 8. Data Cleaning - Duplicates

**Output example:**
```

   Name  Age
0  John   25
1  Jane   28

After drop_duplicates():
```

````python
# Create data with duplicates
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'John', 'Bob', 'Jane'],
    'Age': [25, 28, 25, 30, 28],
    'City': ['NYC', 'LA', 'NYC', 'Chicago', 'LA']
})

print("Original DataFrame with duplicates:")
print(df)

# Check for duplicates
print("\nDuplicate rows:")
print(df.duplicated())

# Drop duplicates
print("\nAfter drop_duplicates():")
print(df.drop_duplicates())

# Drop duplicates based on specific columns
print("\nAfter drop_duplicates(subset=['Name']):")
print(df.drop_duplicates(subset=['Name']))

# Keep last occurrence
print("\nKeep last duplicate:")
print(df.drop_duplicates(keep='last'))
````

## 9. Data Type Conversion

**Output example:**
```
Before conversion:

After astype conversion:
Age       int64
Salary   float64
```

````python
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob'],
    'Age': ['25', '28', '30'],
    'Salary': ['50000', '60000', '75000']
})

print("Original data types:")
print(df.dtypes)

# Convert to numeric
df['Age'] = df['Age'].astype(int)
df['Salary'] = df['Salary'].astype(float)

print("\nAfter conversion:")
print(df.dtypes)
print("\nDataFrame:")
print(df)

# Convert to datetime
df2 = pd.DataFrame({'Date': ['2024-01-01', '2024-01-02']})
df2['Date'] = pd.to_datetime(df2['Date'])
print("\nConverted to datetime:")
print(df2.dtypes)
````

## 10. Filtering and Sorting

**Output example:**
```
Filtered (Age > 26):
   Name  Age  Salary
1  Jane   28   60000
2   Bob   30   75000

Sorted by Salary (descending):
   Name  Age  Salary
2   Bob   30   75000
1  Jane   28   60000
0  John   25   50000
```

````python
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice'],
    'Age': [25, 28, 30, 26],
    'Salary': [50000, 60000, 75000, 55000],
    'Department': ['IT', 'HR', 'IT', 'Finance']
})

# Single condition filter
print("Age > 26:")
print(df[df['Age'] > 26])

# Multiple conditions (AND)
print("\nAge > 25 AND Salary > 55000:")
print(df[(df['Age'] > 25) & (df['Salary'] > 55000)])

# Multiple conditions (OR)
print("\nDepartment = IT OR Age > 28:")
print(df[(df['Department'] == 'IT') | (df['Age'] > 28)])

# Sort by single column
print("\nSorted by Salary (ascending):")
print(df.sort_values('Salary'))

# Sort by multiple columns
print("\nSorted by Department then Salary (descending):")
print(df.sort_values(['Department', 'Salary'], ascending=[True, False]))

# Using isin()
print("\nDepartment in ['IT', 'Finance']:")
print(df[df['Department'].isin(['IT', 'Finance'])])
````

## 12. GroupBy Operations

**Output example:**
```
Sales by Department:
Department
Finance         110000
HR              60000
IT             125000

Average Salary by Department:
Department
Finance     55000
HR          60000
IT          62500
```

````python
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'Finance'],
    'Salary': [50000, 60000, 75000, 55000, 65000],
    'Bonus': [5000, 3000, 7500, 4000, 6500]
})

print("Original DataFrame:")
print(df)

# Group by single column
print("\nTotal Salary by Department:")
print(df.groupby('Department')['Salary'].sum())

# Group by with multiple aggregations
print("\nSalary statistics by Department:")
print(df.groupby('Department')['Salary'].agg(['sum', 'mean', 'count']))

# Group by with multiple aggregations on multiple columns
print("\nMultiple aggregations:")
print(df.groupby('Department').agg({'Salary': 'sum', 'Bonus': 'mean'}))

# Group by with custom function
print("\nTotal compensation by Department:")
print(df.groupby('Department')[['Salary', 'Bonus']].sum().sum(axis=1))
````

## 13. Pivot Tables

**Output example:**
```
Pivot table - Total Sales by Product and Region:
Region      East  North  South  West
Product
A           300    250    350   400
B           400    350    300   350
```

````python
df = pd.DataFrame({
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Region': ['East', 'East', 'West', 'West', 'North', 'North'],
    'Sales': [300, 400, 400, 350, 250, 350],
    'Date': pd.date_range('2024-01-01', periods=6)
})

print("Original DataFrame:")
print(df)

# Simple pivot table
print("\nPivot: Sales by Product and Region:")
pivot = df.pivot_table(values='Sales', index='Product', columns='Region', aggfunc='sum')
print(pivot)

# Pivot with multiple aggregations
print("\nPivot: Mean and Sum:")
pivot2 = df.pivot_table(values='Sales', index='Product', columns='Region', aggfunc=['sum', 'mean'])
print(pivot2)

# Unstack
print("\nUsing unstack (similar to pivot):")
grouped = df.groupby(['Product', 'Region'])['Sales'].sum()
print(grouped.unstack())
````

## 14. Merging and Joining

**Output example:**
```
After merge on 'ID':
   ID  Name  Salary  Department
0   1  John   50000         IT
1   2  Jane   60000         HR
```

````python
# Create two DataFrames
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['John', 'Jane', 'Bob'],
    'Salary': [50000, 60000, 75000]
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Department': ['IT', 'HR', 'IT']
})

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)

# Inner merge
print("\nInner merge on ID:")
merged = pd.merge(df1, df2, on='ID')
print(merged)

# Left merge
df3 = pd.DataFrame({'ID': [1, 2, 4], 'Location': ['NYC', 'LA', 'Chicago']})
print("\nLeft merge (includes all from df1):")
print(pd.merge(df1, df3, on='ID', how='left'))

# Concatenate
df4 = pd.DataFrame({
    'ID': [4, 5],
    'Name': ['Alice', 'Charlie'],
    'Salary': [55000, 65000]
})
print("\nConcatenate vertically:")
print(pd.concat([df1, df4], ignore_index=True))
````

## 15. Apply and Transform

**Output example:**
```
After apply with square function:
0     625
1     784
2     900

After transform (z-score):
0   -1.224745
1   -0.408248
2    1.632993
```

````python
df = pd.DataFrame({
    'Age': [25, 28, 30],
    'Salary': [50000, 60000, 75000]
})

print("Original DataFrame:")
print(df)

# Apply function to column
print("\nApply custom function (square):")
df['Age_squared'] = df['Age'].apply(lambda x: x ** 2)
print(df[['Age', 'Age_squared']])

# Apply to multiple columns
print("\nApply custom function to multiple columns:")
df['Both_squared'] = df[['Age', 'Salary']].apply(lambda row: row['Age']**2 + row['Salary'], axis=1)
print(df[['Age', 'Salary', 'Both_squared']])

# Transform (standardization)
from sklearn.preprocessing import StandardScaler
print("\nTransform (standardization):")
df_copy = df.copy()
df_copy['Salary_standardized'] = (df_copy['Salary'] - df_copy['Salary'].mean()) / df_copy['Salary'].std()
print(df_copy[['Salary', 'Salary_standardized']])
````

## 17. Date and Time Operations

**Output example:**
```
Date column extracted:
2024-01-01    100
2024-01-02    150
2024-01-03    120

Resampled to weekly:
2024-01-07    370
2024-01-14    120
```

````python
# Create time series data
dates = pd.date_range('2024-01-01', periods=10, freq='D')
data = np.random.randint(100, 200, 10)
df = pd.DataFrame({'Date': dates, 'Sales': data})

print("Time series DataFrame:")
print(df)

# Extract date components
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['DayOfWeek'] = df['Date'].dt.dayofweek

print("\nWith date components:")
print(df[['Date', 'Year', 'Month', 'Day', 'DayOfWeek']].head())

# Set date as index
df_ts = df.set_index('Date')
print("\nWith date index:")
print(df_ts.head())

# Resample to weekly
print("\nResampled to weekly (sum):")
print(df_ts['Sales'].resample('W').sum())

# Rolling window
print("\nRolling mean (3-day window):")
print(df_ts['Sales'].rolling(window=3).mean())
````

**Output example:**

````python
df = pd.DataFrame({
    'Product': ['A', 'B', 'A', 'C', 'B', 'A'],
    'Priority': ['High', 'Low', 'Medium', 'High', 'Low', 'Medium']
})

print("Original dtypes:")
print(df.dtypes)

# Convert to categorical
df['Product'] = pd.Categorical(df['Product'])
df['Priority'] = pd.Categorical(
    df['Priority'],
    categories=['Low', 'Medium', 'High'],
    ordered=True
)

print("\nAfter converting to categorical:")
print(df.dtypes)

print("\nValue counts for Priority:")
print(df['Priority'].value_counts().sort_index())

print("\nOrdered categorical (Low < Medium < High):")
print(df[df['Priority'] > 'Low'])
````

## 20. Advanced: Window Functions and Ranking

**Output example:**
```
Cumulative sum by department:
Department  IT   HR  Finance
Cumsum      50000 60000 55000

Rank within group:
Department  IT  HR  Finance
Rank        2   1   1
```

````python
df = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'Finance'],
    'Salary': [50000, 60000, 75000, 55000, 65000]
})

print("Original DataFrame:")
print(df)

# Cumulative sum by department
print("\nCumulative sum by Department:")
df['Cumsum'] = df.groupby('Department')['Salary'].cumsum()
print(df[['Name', 'Department', 'Salary', 'Cumsum']])

# Rank within group
print("\nRank within Department (by Salary):")
df['Rank'] = df.groupby('Department')['Salary'].rank(ascending=False)
print(df[['Name', 'Department', 'Salary', 'Rank']])

# Percent change
print("\nPercent change (if ordered by salary):")
df_sorted = df.sort_values('Salary')
df_sorted['PctChange'] = df_sorted['Salary'].pct_change() * 100
print(df_sorted[['Name', 'Salary', 'PctChange']])
````

## 21. Performance Tips & Best Practices

**Key optimization strategies:**
1. Use `copy=False` when possible to avoid unnecessary copies
2. Use categorical data for repeated strings
3. Use `inplace=True` for memory efficiency
4. Vectorize operations instead of loops
5. Use `query()` for faster filtering on large datasets

````python
import time

# Create large dataset
df_large = pd.DataFrame({
    'ID': range(100000),
    'Value': np.random.randint(1, 1000, 100000)
})

# Loop method (slow)
start = time.time()
result_loop = [x * 2 for x in df_large['Value']]
time_loop = time.time() - start

# Vectorized method (fast)
start = time.time()
result_vec = df_large['Value'] * 2
time_vec = time.time() - start

print(f"Loop time: {time_loop:.6f}s")
print(f"Vectorized time: {time_vec:.6f}s")
print(f"Speedup: {time_loop/time_vec:.0f}x faster!")

# Using query for filtering
start = time.time()
df_filtered_loop = df_large[df_large['Value'] > 500]
time_filter1 = time.time() - start

start = time.time()
df_filtered_query = df_large.query('Value > 500')
time_filter2 = time.time() - start

print(f"\nFilter with boolean: {time_filter1:.6f}s")
print(f"Filter with query: {time_filter2:.6f}s")
````

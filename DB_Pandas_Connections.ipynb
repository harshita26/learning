# Connecting to MySQL and PostgreSQL with Pandas

This notebook demonstrates how to connect to MySQL using PyMySQL and to PostgreSQL using Psycopg2, and how to read data from both databases into pandas DataFrames.

## 2. Import Libraries

Import the required libraries for database connections and data analysis.

```python
import pandas as pd
import pymysql
import psycopg2
```

## 3. Connect to MySQL Database with PyMySQL

Set up a connection to a MySQL database using PyMySQL. Replace the example parameters with your actual credentials.

```python
# Example connection parameters (replace with your own)
mysql_config = {
    'host': 'localhost',
    'user': 'your_mysql_user',
    'password': 'your_mysql_password',
    'database': 'your_database',
    'port': 3306
}

# Establish connection
mysql_conn = pymysql.connect(**mysql_config)
print('Connected to MySQL!')
```

## 4. Read Data from MySQL into Pandas DataFrame

Use pandas to execute a SQL query and load the results into a DataFrame.

```python
# Example SQL query
mysql_query = "SELECT * FROM your_table LIMIT 5;"

# Read data into DataFrame
mysql_df = pd.read_sql(mysql_query, mysql_conn)
print(mysql_df.head())
```

## 5. Connect to PostgreSQL Database with Psycopg2

Set up a connection to a PostgreSQL database using Psycopg2. Replace the example parameters with your actual credentials.

```python
# Example connection parameters (replace with your own)
pg_config = {
    'host': 'localhost',
    'user': 'your_pg_user',
    'password': 'your_pg_password',
    'dbname': 'your_database',
    'port': 5432
}

# Establish connection
pg_conn = psycopg2.connect(**pg_config)
print('Connected to PostgreSQL!')
```

## 6. Read Data from PostgreSQL into Pandas DataFrame

Use pandas to execute a SQL query and load the results into a DataFrame.

```python
# Example SQL query
pg_query = "SELECT * FROM your_table LIMIT 5;"

# Read data into DataFrame
pg_df = pd.read_sql(pg_query, pg_conn)
print(pg_df.head())
```
